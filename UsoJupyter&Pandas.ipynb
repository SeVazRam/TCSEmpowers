## Operadores y expresiones

Los **operadores** permiten realizar cálculos, comparar valores y combinar condiciones. Una **expresión** es una combinación de valores, variables y operadores que produce un resultado.

### 1. Operadores aritméticos

| Operador | Descripción | Ejemplo |
|---|---|---|
| `+` | Suma | `5 + 2` → `7` |
| `-` | Resta | `5 - 2` → `3` |
| `*` | Multiplicación | `5 * 2` → `10` |
| `/` | División | `5 / 2` → `2.5` |
| `//` | División entera | `5 // 2` → `2` |
| `%` | Módulo (residuo) | `5 % 2` → `1` |
| `**` | Potencia | `5 ** 2` → `25` |

### 2. Operadores de comparación

Comparan dos valores y devuelven un resultado booleano: `True` o `False`.

- `==` igual que
- `!=` diferente de
- `<` menor que
- `<=` menor o igual que
- `>` mayor que
- `>=` mayor o igual que

**Importante:** `=` asigna un valor; `==` compara dos valores.

### 3. Operadores lógicos

Permiten combinar o invertir condiciones.

- `and`: devuelve `True` cuando ambas condiciones son verdaderas.
- `or`: devuelve `True` cuando al menos una condición es verdadera.
- `not`: invierte el resultado lógico.

### 4. Orden de las operaciones

Python sigue reglas de precedencia para evaluar expresiones. Los paréntesis `()` permiten indicar explícitamente qué operaciones deben realizarse primero y hacen el código más fácil de leer.

````python
subtotal = 3 * 25.50

es_mayor = subtotal > 50
con_descuento = es_mayor and subtotal >= 75

print(subtotal, es_mayor, con_descuento)
````

## Colecciones: listas (`list`)

Una **lista** es una colección que permite almacenar varios elementos en una sola variable. Las listas mantienen el orden de sus elementos y pueden modificarse después de crearse.

### Características principales

- **Orden:** los elementos conservan su posición.
- **Índices:** comienzan en `0`, por lo que el primer elemento está en la posición `0`.
- **`append(elemento)`:** agrega un elemento al final de la lista.
- **`len(lista)`:** devuelve la cantidad de elementos.
- **Mutabilidad:** podemos modificar, agregar o eliminar elementos.

### Operaciones básicas

| Operación | Descripción |
|---|---|
| `lista[0]` | Accede al primer elemento |
| `lista[-1]` | Accede al último elemento |
| `lista.append(x)` | Agrega `x` al final |
| `len(lista)` | Devuelve el número de elementos |
| `lista[1:3]` | Obtiene elementos desde el índice 1 hasta antes del 3 |
| `lista.sort()` | Ordena la lista en el mismo objeto |


````python
calificaciones = [8.5, 6.0, 9.2, 4.8, 7.5]

# Calcular el promedio
promedio = sum(calificaciones) / len(calificaciones)

# Identificar las calificaciones aprobatorias
aprobadas = []

for calificacion in calificaciones:
    if calificacion >= 6:
        aprobadas.append(calificacion)

# Agregar una nueva calificación
calificaciones.append(9.0)

print("Calificaciones:", calificaciones)
print("Cantidad:", len(calificaciones))
print("Promedio original:", round(promedio, 2))
print("Calificaciones aprobatorias:", aprobadas)
````

## Estructuras de control en Python

Las **estructuras de control** permiten modificar el flujo de ejecución de un programa. Con ellas podemos tomar decisiones, repetir instrucciones y procesar colecciones de datos.

### 1. Condicionales: `if`, `elif` y `else`

Permiten ejecutar diferentes bloques de código según se cumplan determinadas condiciones.

**Sintaxis:**

```python
if condicion_1:
    # Se ejecuta si condicion_1 es verdadera
elif condicion_2:
    # Se ejecuta si condicion_1 es falsa
    # y condicion_2 es verdadera
else:
    # Se ejecuta si ninguna condición anterior se cumple
```

- `if`: evalúa la primera condición.
- `elif`: evalúa otra condición si las anteriores fueron falsas.
- `else`: ejecuta una alternativa cuando ninguna condición anterior se cumple.
- Los dos puntos `:` indican el inicio del bloque.
- La indentación define qué instrucciones pertenecen a cada bloque.

### 2. Bucle `for`

Se utiliza para recorrer elementos de una colección o repetir instrucciones un número determinado de veces.

**Sintaxis:**

```python
for elemento in coleccion:
    # Instrucciones que se repiten
```

Ejemplo de estructura con `range()`:

```python
for numero in range(5):
    print(numero)
```

`range(5)` genera los valores del 0 al 4.

### 3. Bucle `while`

Repite un bloque de código mientras una condición sea verdadera. Es útil cuando no conocemos de antemano cuántas iteraciones serán necesarias.

**Sintaxis:**

```python
while condicion:
    # Instrucciones que se repiten
    # Actualizar el estado para evitar un bucle infinito
```

**Importante:** si la condición nunca se vuelve falsa, el bucle puede ejecutarse indefinidamente.

### Resumen

| Estructura | Propósito |
|---|---|
| `if / elif / else` | Tomar decisiones |
| `for` | Recorrer una colección o secuencia |
| `while` | Repetir mientras se cumpla una condición |

````python
# Sistema de análisis de calificaciones
estudiantes = [
    {"nombre": "Ana", "calificacion": 95},
    {"nombre": "Luis", "calificacion": 72},
    {"nombre": "Marta", "calificacion": 58},
    {"nombre": "Carlos", "calificacion": 84},
    {"nombre": "Sofia", "calificacion": 40}
]

resultados = []
indice = 0
limite = len(estudiantes)

# WHILE: procesa registros mientras queden estudiantes
while indice < limite:

    estudiante = estudiantes[indice]
    nombre = estudiante["nombre"]
    calificacion = estudiante["calificacion"]

    # IF / ELIF / ELSE: clasifica el rendimiento
    if calificacion >= 90:
        nivel = "Excelente"
    elif calificacion >= 80:
        nivel = "Muy bueno"
    elif calificacion >= 60:
        nivel = "Aprobado"
    else:
        nivel = "Reprobado"

    # Guarda el resultado de cada estudiante
    resultados.append({
        "nombre": nombre,
        "calificacion": calificacion,
        "nivel": nivel
    })

    indice += 1  # Avanza al siguiente estudiante

# FOR: muestra el reporte final
print("REPORTE DE CALIFICACIONES")
print("-" * 40)

for resultado in resultados:
    print(
        f"{resultado['nombre']:<10} | "
        f"{resultado['calificacion']:>3} | "
        f"{resultado['nivel']}"
    )

# FOR + IF: calcula estadísticas
aprobados = 0
excelentes = 0
suma_calificaciones = 0

for resultado in resultados:
    suma_calificaciones += resultado["calificacion"]

    if resultado["calificacion"] >= 60:
        aprobados += 1

    if resultado["calificacion"] >= 90:
        excelentes += 1

promedio = suma_calificaciones / len(resultados)

print("-" * 40)
print(f"Estudiantes procesados: {len(resultados)}")
print(f"Aprobados: {aprobados}")
print(f"Excelentes: {excelentes}")
print(f"Promedio general: {promedio:.2f}")
````

## Diccionarios en Python (`dict`)

Un **diccionario** es una estructura de datos que permite almacenar y organizar información en pares de **clave y valor** (*key-value pairs*). Cada clave sirve como identificador para acceder al valor asociado.

Los diccionarios son útiles para representar información estructurada, como los datos de una persona, las características de un producto o la configuración de una aplicación.

### Características principales

- **Claves únicas:** cada clave aparece una sola vez; si se asigna nuevamente, su valor se actualiza.
- **Acceso mediante claves:** los valores se consultan utilizando sus claves, en lugar de índices numéricos.
- **Mutabilidad:** es posible agregar, modificar y eliminar elementos después de crear el diccionario.
- **Tipos de datos flexibles:** las claves suelen ser cadenas de texto, aunque también pueden ser otros tipos inmutables. Los valores pueden ser de distintos tipos, como números, cadenas, listas, booleanos u otros diccionarios.
- **Orden de inserción:** Python conserva el orden en que se agregan los elementos.

### Sintaxis

Los diccionarios se crean utilizando llaves `{}`, separando cada clave de su valor mediante dos puntos (`:`) y los pares mediante comas.

```python
diccionario = {
    "clave1": "valor1",
    "clave2": "valor2",
    "clave3": "valor3"
}
```

### Operaciones básicas

- **Consultar:** obtener un valor mediante su clave.
- **Agregar:** incorporar un nuevo par de clave y valor.
- **Modificar:** cambiar el valor asociado a una clave existente.
- **Eliminar:** quitar un elemento del diccionario.
- **Recorrer:** iterar sobre sus claves, valores o pares mediante ciclos.

En Python, el tipo de dato de un diccionario es `dict`.

Los diccionarios permiten representar información de manera clara y acceder a ella mediante identificadores significativos, lo que facilita la lectura y el mantenimiento del código.

````python
#Ejemplo basico
persona = {
    "nombre": "Ana",
    "edad": 25,
    "ciudad": "Ciudad de México"
}

# Acceder a un valor
print(persona["nombre"])

# Modificar un valor
persona["edad"] = 26

# Agregar una nueva clave
persona["profesion"] = "Ingeniera"

print(persona)
````

````python
#Ejemplo intermedio
inventario = {
    "Laptop": 15000,
    "Mouse": 350,
    "Teclado": 800,
    "Monitor": 4200
}

total = 0

for producto, precio in inventario.items():
    print(f"{producto}: ${precio}")
    total += precio

print(f"Valor total: ${total}")
````

````python
#Ejemplo avanzado
ventas = [
    {"producto": "Laptop", "cantidad": 3, "precio": 15000},
    {"producto": "Mouse", "cantidad": 10, "precio": 350},
    {"producto": "Monitor", "cantidad": 4, "precio": 4200},
    {"producto": "Teclado", "cantidad": 6, "precio": 800}
]

total_ventas = 0
producto_mayor_venta = ""
mayor_venta = 0

for venta in ventas:
    subtotal = venta["cantidad"] * venta["precio"]
    total_ventas += subtotal

    print(f"{venta['producto']}: ${subtotal:,.2f}")

    if subtotal > mayor_venta:
        mayor_venta = subtotal
        producto_mayor_venta = venta["producto"]

print("-" * 35)
print(f"Total de ventas: ${total_ventas:,.2f}")
print(f"Mayor venta: {producto_mayor_venta}")
print(f"Importe mayor: ${mayor_venta:,.2f}")
````

## Funciones en Python (`def`)

Una **función** es un bloque de código reutilizable que realiza una tarea específica. Permite organizar el código, evitar repeticiones y facilitar su mantenimiento.

### Sintaxis

```python

    # Instrucciones
    return resultado
```

- **`def`:** palabra clave para definir una función.
- **Nombre:** identificador de la función.
- **Parámetros:** datos que recibe la función (opcionales).
- **`return`:** devuelve un resultado (opcional).

### Ejemplo

```python

    return a + b

resultado = sumar(5, 3)
print(resultado)
```

**Salida:**
```text
8
```

Para ejecutar una función, debemos **llamarla por su nombre** y proporcionar los argumentos necesarios.

Las funciones pueden recibir parámetros, devolver resultados o simplemente ejecutar instrucciones.

````python

def mostrar_mensaje():
    print("Bienvenido al curso de Python")
    print("Aprenderemos a utilizar funciones.")

mostrar_mensaje()
````

````python

def calcular_promedio(nota1, nota2, nota3):
    promedio = (nota1 + nota2 + nota3) / 3
    return promedio

resultado = calcular_promedio(90, 85, 95)

print(f"Promedio: {resultado:.2f}")
````

````python

def analizar_ventas(ventas):
    total = 0
    producto_mayor = ""
    mayor_importe = 0

    for venta in ventas:
        importe = venta["cantidad"] * venta["precio"]
        total += importe

        if importe > mayor_importe:
            mayor_importe = importe
            producto_mayor = venta["producto"]

    return {
        "total_ventas": total,
        "producto_mayor": producto_mayor,
        "mayor_importe": mayor_importe
    }


ventas = [
    {"producto": "Laptop", "cantidad": 3, "precio": 15000},
    {"producto": "Mouse", "cantidad": 10, "precio": 350},
    {"producto": "Monitor", "cantidad": 4, "precio": 4200}
]

resumen = analizar_ventas(ventas)

print(f"Total de ventas: ${resumen['total_ventas']:,.2f}")
print(f"Producto destacado: {resumen['producto_mayor']}")
print(f"Mayor importe: ${resumen['mayor_importe']:,.2f}")
````

```python
df.groupby("Producto")["Importe"].sum()
```

```python
pd.pivot_table(
    df,
    values="Importe",
    index="Producto",
    columns="Categoria",
    aggfunc="sum",
    fill_value=0
)
```

## 2. Map y Apply

Permiten **transformar datos mediante funciones**.

- `map()` aplica una correspondencia o función a cada elemento de una `Series`, por ejemplo, convertir códigos en nombres.
- `apply()` permite aplicar una función a una columna o a cada fila o columna de un `DataFrame`.

```python
# Reemplazar códigos por nombres
df["Categoria"] = df["Categoria"].map({
    "A": "Accesorios",
    "C": "Computadoras"
})

# Aplicar una operación a una columna
df["Precio_con_IVA"] = df["Precio"].apply(
    lambda precio: precio * 1.16
)
```

## 3. Merge

Permite **combinar dos DataFrames utilizando una columna en común**, de forma similar a las uniones (`JOIN`) de SQL.

```python
resultado = pd.merge(
    ventas,
    productos,
    on="Producto",
    how="inner"
)
```

El parámetro `how` define el tipo de unión: `inner`, `left`, `right` u `outer`.

## 4. Append y concatenación

Permiten **unir datos de varias tablas**. El método `DataFrame.append()` ya no está disponible en versiones modernas de Pandas; se utiliza `pd.concat()`.

```python
# Unir dos DataFrames por filas
resultado = pd.concat(
    [df_enero, df_febrero],
    ignore_index=True
)
```

`ignore_index=True` genera un nuevo índice consecutivo para las filas combinadas.

## 5. Exportar datos

Pandas permite guardar los resultados en archivos para compartirlos o utilizarlos en otras aplicaciones.

```python
# Exportar a CSV
df.to_csv("resultados.csv", index=False)

# Exportar a Excel
df.to_excel("resultados.xlsx", index=False)
```

El parámetro `index=False` evita exportar el índice como una columna adicional.

Estas operaciones permiten resumir información, transformar valores, combinar fuentes de datos y guardar los resultados de un análisis.

````python
### Datos de ejemplo

datos_basico = {
    "producto": ["Teclado", "Mouse", "Monitor", "Webcam", "Audífonos"],
    "categoria": ["Accesorios", "Accesorios", "Pantallas", "Accesorios", "Audio"],
    "precio": [700, 250, 3200, 800, 900],
    "unidades_vendidas": [12, 25, 7, 10, 18]
}

df_basico = pd.DataFrame(datos_basico)

display(df_basico.head())
df_basico["ingresos"] = df_basico["precio"] * df_basico["unidades_vendidas"]
display(df_basico[df_basico["ingresos"] >= 1000].sort_values("ingresos", ascending=False))
````

````python
resumen = (
    df_basico.groupby("categoria")
      .agg(
          ingresos_totales=("ingresos", "sum"),
          unidades_totales=("unidades_vendidas", "sum")
      )
      .reset_index()
)
display(resumen.sort_values("ingresos_totales", ascending=False))
top = resumen.loc[resumen["ingresos_totales"].idxmax(), "categoria"]
print("Categoría con más ingresos:", top)
````

````python
# Crear una tabla dinámica
tabla_dinamica = pd.pivot_table(
    df_basico,
    values="ingresos",
    index="categoria",
    columns="producto",
    aggfunc="sum",
    fill_value=0
)

display(tabla_dinamica)
````

````python
# Diccionario para traducir las categorías
mapa_categorias = {
    "Accesorios": "Complementos",
    "Pantallas": "Monitores",
    "Audio": "Sonido"
}

# Crear una nueva columna con las categorías traducidas
df_basico["categoria_descripcion"] = (
    df_basico["categoria"].map(mapa_categorias)
)

display(df_basico)
````

````python
df_basico["precio_con_iva"] = df_basico["precio"].apply(
    lambda precio: precio * 1.16
)

display(df_basico[["producto", "precio", "precio_con_iva"]])
````

````python
datos_proveedores = {
    "producto": ["Teclado", "Mouse", "Monitor", "Webcam", "Audífonos"],
    "proveedor": ["Logitech", "HP", "Samsung", "Logitech", "Sony"]
}

df_proveedores = pd.DataFrame(datos_proveedores)

display(df_proveedores)
````

````python
datos_nuevos = {
    "producto": ["Micrófono", "Bocinas"],
    "categoria": ["Audio", "Audio"],
    "precio": [1200, 1500],
    "unidades_vendidas": [8, 6]
}

df_nuevos = pd.DataFrame(datos_nuevos)

df_concatenado = pd.concat(
    [df_basico, df_nuevos],
    ignore_index=True
)

display(df_concatenado)
````
